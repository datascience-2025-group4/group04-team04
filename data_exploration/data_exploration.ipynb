```python
import matplotlib.pyplot as plt
import pandas as pd


#Kopie des annotated datensatzes erstellen
ab_ag_annotated_copy = pd.read_csv('data/ab_ag_annotated.tsv', sep='\t')
#anzahl der einträge vor der Löschung zur Kontrolle
count_beforedeletion = len(ab_ag_annotated_copy)
#löschung der einträge mit identischer sequenz
ab_ag_uniquesequences = ab_ag_annotated_copy.drop_duplicates(subset=['pdb', 'CDR_H1','CDR_H2', 'CDR_L1', 'CDR_L2', 'CDR_L3'] )


#histogramm mit antigen häufigkeiten


#Häufigkeiten der antigen_name
antigen_counts = ab_ag_uniquesequences["antigen_name"].value_counts()

#nur die Top 10 anzeigen
top_n = 10
antigen_counts = antigen_counts.head(top_n)

plt.figure(figsize=(12, 6))  # Breite x Höhe von barplot

#Barplot => x=Antigen-Namen, y=Häufigkeit
plt.bar(antigen_counts.index, antigen_counts.values)

plt.xlabel("Antigen Name")
plt.ylabel("Anzahl der Antikörper (Häufigkeit)")
plt.title("Häufigkeit der Antigen-Namen in ab_ag_uniquesequences")
plt.xticks(rotation=45, ha='right')  #x-Achsen-Labels drehen für bessere Lesbarkeit

plt.tight_layout()  #Optimiert Abstände
plt.show()
```

```python
import pandas as pd

ab_ag_annotated = pd.read_csv('data/ab_ag_annotated.tsv', sep='\t')
# Häufigkeiten der antigen_name nur in ab_ag_annotated
antigen_counts = ab_ag_annotated["antigen_name"].value_counts()

# nur Antigene anzeigen, die >=10-mal vorkommen
antigen_counts = antigen_counts[antigen_counts >= 10]

plt.figure(figsize=(12, 6))  # Breite x Höhe von Barplot
plt.bar(antigen_counts.index, antigen_counts.values)

plt.xlabel("Antigen Name")
plt.ylabel("Anzahl der Antikörper (Häufigkeit)")
plt.title("Häufigkeit der Antigen-Namen in ab_ag_annotated (>=10)")
plt.xticks(rotation=45, ha='right')  # X-Achsen-Labels drehen für bessere Lesbarkeit

plt.tight_layout()  # Optimiert Abstände
plt.show()
```
