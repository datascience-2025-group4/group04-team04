```python
from data.sabdab_downloader import getsequence
import pandas as pd
import numpy as np
import os

df = pd.read_csv("data/ab_ag.tsv", sep = "\t")
```

```python
fab_lists = df[["pdb", "Hchain", "Lchain", "model", "antigen_name", "antigen_species"]].values.tolist()
df_rows = []

for fab_list in fab_lists:
    pdb = fab_list[0]
    hchain = fab_list[1]
    lchain = fab_list[2]
    model = fab_list[3]
    antigen_name = fab_list[4]
    antigen_species = fab_list[5]

    VH_filename = f"{pdb}_{fab_list[1]}_VH.fa"
    VL_filename = f"{pdb}_{fab_list[2]}_VL.fa"

    pdb_path = f"data/download/v_seq/{pdb}"
    

    if not os.path.exists(pdb_path):
        os.makedirs(pdb_path)

    try:
        getsequence(pdb, [fab_list], pdb_path)
        with open(os.path.join(pdb_path, VH_filename)) as vh_fasta:
            vh_seq = vh_fasta.readlines()[3].replace("-", "")
        with open(os.path.join(pdb_path, VL_filename)) as vl_fasta:
            vl_seq = vl_fasta.readlines()[3].replace("-", "")
        df_row = [pdb, hchain, lchain, model, antigen_name, antigen_species, vh_seq, vl_seq]
    
    except Exception as e:
        print(f"Fehler bei {pdb}: {e}")
        df_row = [pdb, hchain, lchain, model, antigen_name, antigen_species, np.nan, np.nan]
    
    df_rows.append(df_row)



ab_ag_vseqs = (
    pd.DataFrame(df_rows, columns= ["pdb", "Hchain", "Lchain", "model", "antigen_name", "antigen_species", "VH", "VL"])
    .dropna(subset = ["VH", "VL"])
    #.drop_duplicates(subset = ["VH", "VL"])
    .reset_index(drop = True)
)


ab_ag_vseqs.to_csv('data/ab_ag_vseqs.tsv', sep='\t', index=False)
```
