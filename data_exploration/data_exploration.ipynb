```python
import pandas as pd
import matplotlib.pyplot as plt
seq_regions = ['SEQ_H1', 'SEQ_H2', 'SEQ_L1', 'SEQ_L2', 'SEQ_L3']
cf_regions = ['CF_H1', 'CF_H2', 'CF_L1', 'CF_L2', 'CF_L3']

df = (
    pd.read_csv("data/ab_ag_scalop.tsv", sep="\t")
    .dropna(subset=seq_regions)
    .dropna(subset=cf_regions)
    .drop_duplicates(subset=seq_regions) 
)
antigen_counts = df["antigen_name"].value_counts()

# nur Antigene anzeigen, die >=10-mal vorkommen
antigen_counts = antigen_counts[antigen_counts >= 10]

plt.figure(figsize=(12, 6))  # Breite x Höhe von Barplot
plt.bar(antigen_counts.index, antigen_counts.values)

plt.xlabel("Antigen Name")
plt.ylabel("Anzahl der Antikörper (Häufigkeit)")
plt.title("Häufigkeit der Antigen-Namen in ab_ag_scalop (>=10)")
plt.xticks(rotation=45, ha='right')  # X-Achsen-Labels drehen für bessere Lesbarkeit

plt.tight_layout()  # Optimiert Abstände
plt.show()
```
